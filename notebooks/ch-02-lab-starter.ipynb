## Step 2: get one line's value

Before writing the whole function, make sure you can handle a single line. Two
things matter here.

First, pulling numbers out of one dictionary: `one_line["quantity"]` gives you the
quantity, and so on.

Second, the awkward part. Not every line has a `discount_pct` key, and asking a
dictionary for a key it does not have raises a `KeyError`. The chapter gives you the
tool for this in the dictionary section: `one_line.get("discount_pct", 0.0)` returns
the discount when the key is there and `0.0` when it is not. That is exactly the
fallback you want, since a line with no discount has a discount of zero.

This one is written for you, because the point of the lab is what comes next.


````python
def value_of(one_line):
    """Post-discount value of a single order line."""
    return line_total(
        one_line["quantity"],
        one_line["unit_price"],
        one_line.get("discount_pct", 0.0),   # 0.0 when the key is absent
    )


value_of(my_order[0])   # try it on your first line
````
