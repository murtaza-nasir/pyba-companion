## Step 2: make your working copy

The cell below copies the starter script to `lab_6_dashboard_starter.py` in your
session, so that your edits stay separate from the pristine copy in the
repository.

After running it, open the file in Colab's editor: click the **folder icon** in
the left sidebar, find `lab_6_dashboard_starter.py`, and double-click it. It opens
in an editor pane on the right. Edit there, and press **Ctrl+S** to save.

````python
import shutil, os

SRC = "pyba-companion/dashboards/lab_6_dashboard_starter.py"
DST = "lab_6_dashboard_starter.py"

if not os.path.exists(DST):
    shutil.copy(SRC, DST)
    print(f"copied to {DST}")
else:
    print(f"{DST} already exists; keeping your edited version")

print(open(DST).read()[:400] + " ...")
````

## Step 3: launch it

This is the same launcher Chapter 6 uses, pointed at your working copy. It starts
the Streamlit server in this Colab session and opens the dashboard in a new browser
tab through Colab's port proxy.

Take your screenshot from that tab. To stop the app, interrupt this cell or restart
the runtime. **Re-run this cell after every edit** to see your changes.

````python
import os, subprocess, sys, time

if "google.colab" in sys.modules:
    env = {**os.environ, "PYTHONPATH": "pyba-companion"}   # so the app finds pyba
    subprocess.Popen(
        ["streamlit", "run", "lab_6_dashboard_starter.py",
         "--server.headless=true", "--server.port=8501",
         "--server.enableCORS=false", "--server.enableXsrfProtection=false"],
        env=env)
    time.sleep(8)                                          # give the server a moment
    from google.colab import output
    output.serve_kernel_port_as_window(8501)               # prints a link; click it
else:
    print("Not on Colab. Run this instead, from the repository root:")
    print("  streamlit run dashboards/lab_6_dashboard_starter.py")
````
