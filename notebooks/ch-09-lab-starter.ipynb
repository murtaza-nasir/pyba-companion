## Recap: the Chapter 9 setup

The recap cell contains the chapter's code. `X` is the feature table with 20
columns, `y` is the `churned` label, and 262 of the 872 customers are held out
for testing. `full_model` is the chapter's logistic pipeline on all 20
features. Its test accuracy and recall are printed so that you can compare your
model with them in Step 4.

````python
# --- Chapter 9 recap: run this cell first ---
import pandas as pd

from pyba import DATA_DIR
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, confusion_matrix

snap = pd.read_csv(DATA_DIR / "pw_customer_snapshot.csv")
X = pd.get_dummies(snap.drop(columns=["customer_id", "pilot", "churned"]),
                   columns=["business_type", "region"])
y = snap["churned"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, stratify=y, random_state=0)

full_model = make_pipeline(StandardScaler(), LogisticRegression(max_iter=2000))
full_model.fit(X_train, y_train)

full_pred = full_model.predict(X_test)
caught = ((full_pred == 1) & (y_test == 1)).sum()     # churners the full model flagged
print(f"full model: test accuracy {accuracy_score(y_test, full_pred):.3f}, "
      f"recall {caught / (y_test == 1).sum():.3f}")
````
